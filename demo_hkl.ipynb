```python
import math
import ad_hoc_diffractometer as ahd
import hklpy2
import hklpy2.user
```

```python
ahd.register_geometry_file("./diffractometer-geometry.yml", name="cdi-geometry")
ahd.list_geometries()
```

```python
hklpy2.SI_LATTICE_PARAMETER
```

```python
reals = ["mu", "chi", "phi", "omega2", "chi2", "gamma", "delta"]
diff = hklpy2.creator(
    name="cdi-diff",
    geometry="cdi-geometry",
    solver="ad_hoc",
    reals=reals
)
diff.beam.wavelength.put(1.54)
diff.core.extras = {"n_hat": (1, 1, 1)}
hklpy2.user.add_sample("silicon", a=hklpy2.SI_LATTICE_PARAMETER)
hklpy2.user.set_diffractometer(diff)

theta = math.degrees(math.asin((4*1.54)/(2*hklpy2.SI_LATTICE_PARAMETER)))
tth = 2 * theta

r1 = hklpy2.user.setor(
    4,
    0,
    0,
    mu=0,
    chi=0,
    phi=0,
    omega2=theta,
    chi2=0,
    gamma=tth,
    delta=0,
)

r2 = hklpy2.user.setor(
    0,
    4,
    0,
    mu=0,
    chi=0,
    phi=90,
    omega2=theta,
    chi2=0,
    gamma=tth,
    delta=0,
)

hklpy2.user.calc_UB(r1, r2)
```
